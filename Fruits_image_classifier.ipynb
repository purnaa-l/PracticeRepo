```python
import tensorflow as tf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow import keras
from tensorflow.keras import layers
```

```python
data_train_path='/content/drive/MyDrive/Fruits_Vegetables/train'
data_test_path='/content/drive/MyDrive/Fruits_Vegetables/test'
data_validation_path='/content/drive/MyDrive/Fruits_Vegetables/validation'
```

```python
img_width=180
img_height=180

data_train=tf.keras.utils.image_dataset_from_directory(data_train_path, shuffle=True, image_size=(180, 180), batch_size=32, validation_split=False)
```

```python
data_train.class_names
```

```python
data_cat=data_train.class_names
data_val=tf.keras.utils.image_dataset_from_directory(data_validation_path, shuffle=False, image_size=(180, 180), batch_size=32, validation_split=False)
from tensorflow.keras.models import Sequential
model=Sequential([
    layers.Rescaling(1./255), layers.Conv2D(16, 3, padding='same', activation='relu'), layers.MaxPooling2D(), layers.Conv2D(32, 3, padding='same', activation='relu'),
    layers.MaxPooling2D(), layers.Conv2D(64, 3, padding='same', activation='relu'),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dropout(0.2), layers.Dense(128), layers.Dense(len(data_cat))

])
```

```python
model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])
```

```python
history = model.fit(data_train, validation_data=data_val, epochs=25)
```

```python
import tensorflow as tf
from tensorflow.keras.utils import load_img, img_to_array
from tensorflow.keras.utils import array_to_img
import numpy as np

img_width, img_height = 180, 180
class_names = data_cat  # from your training dataset

def load_and_preprocess_image(img_path):
    # Load image and resize
    img = load_img(img_path, target_size=(img_width, img_height))

    # Convert to array
    img_array = img_to_array(img)  # shape (180, 180, 3)

    # Add batch dimension
    img_array = tf.expand_dims(img_array, 0)  # shape (1, 180, 180, 3)

    # No normalization needed because model has Rescaling layer
    return img_array

img_path = 'Image_6.jpg'  # Your test image path

# Load and preprocess image
img = load_and_preprocess_image(img_path)

# Make prediction (logits)
logits = model.predict(img)

# Apply softmax to get probabilities
probabilities = tf.nn.softmax(logits[0]).numpy()

# Get predicted class and confidence
predicted_index = np.argmax(probabilities)
predicted_class = class_names[predicted_index]
confidence = probabilities[predicted_index]

print(f"Predicted class: {predicted_class}")
print(f"Confidence: {confidence:.2%}")

img_display = array_to_img(img[0])
img_display.show()  # To display
```
